```python
# MoletInterface object
sim = MoletInterface(molet_home=MOLET_HOME, conda_env=CONDA_ENV)

# ----------------------------------------------------------------------------
# ------------------------ STEP 1: Cosmological model ------------------------
# ----------------------------------------------------------------------------

sim.set_cosmology(H0=67.7, Wm0=0.309)  # Wm0 is Omega_m at z=0


# ----------------------------------------------------------------------------
# --------------- STEP 2: Instruments, observing epochs, noise ---------------
# ----------------------------------------------------------------------------

# Set the instrument(s)
obs_time = [
    1200.00,1206.99,1214.01,1225.01,1232.98,1239.95,1247.00,1250.95,1253.97,1257.92,1261.96,1266.90,1269.90,1274.90,1277.93,1278.95,1280.92,1281.88,1285.86,1292.90,1299.81,1303.76,1311.77,1314.92,1318.80,1322.79,1325.81,1339.76,1340.79,1342.77,1343.80,1347.66,1350.69,1358.71,1361.70,1361.74,1363.68,1364.68,1367.70,1369.75,1370.77,1372.67,1372.82,1373.71,1375.66,1380.74,1383.69,1389.67,1394.66,1396.70,1400.66,1404.69,1407.68,1412.66,1412.68,1414.68,1416.68,1560.00,1564.00,1567.00,1568.97,1573.97,1579.99,1591.94,1592.95,1592.95,1592.95,1593.97,1594.91,1594.92,1594.94,1595.91,1596.06,1596.92,1596.95,1597.93,1598.98,1599.94,1600.96,1604.82,1605.87,1605.99,1606.90,1607.85,1613.93,1616.02,1617.01,1617.93,1618.96,1619.94,1620.98,1620.99,1621.93,1623.02,1624.00,1626.00,1628.97,1632.00,1633.96,1639.96,1640.98,1641.94,1642.89,1643.81,1645.79,1646.77,1647.82,1649.70,1653.82,1653.94,1655.91,1657.86,1659.90,1661.76,1661.84,1667.76,1670.84,1673.79,1675.79,1678.99,1680.72,1683.80,1684.66,1685.65,1685.93,1687.74,1687.87,1689.65,1689.79,1694.70,1699.65,1700.66,1700.73,1702.61,1702.82,1703.62,1704.66,1706.64,1707.63,1708.84,1709.63,1710.63,1712.60,1712.73,1713.64,1713.73,1714.66,1714.73,1717.74,1721.65,1723.71,1728.65,1732.76,1733.60,1734.73,1736.76,1739.70,1740.70,1753.71,1780.65,1781.68,1789.66,1794.65,1907.03,1911.93,1920.93,1925.88,1927.00,1928.03,1930.01,1930.89,1933.97,1938.00,1939.00,1942.04,1943.95,1945.01,1946.03,1946.96,1952.96,1958.93,1960.95,1967.98,1968.87,1969.95,1976.96,1982.87,1983.91,1986.89,1987.93,1994.82,1995.97,1999.03,1999.97,2002.04,2002.99,2007.02,2012.03,2013.89,2014.71,2016.91,2017.86,2018.72,2023.81,2024.76,2024.82,2025.79,2027.79,2029.78,2029.86,2030.82,2031.84,2032.80,2037.77,2038.92,2039.79,2039.91,2040.81,2042.77,2043.86,2045.81,2051.81,2052.78,2054.90,2055.86,2056.83,2057.78,2058.79,2059.73,2060.76,2061.74,2063.72,2063.75,2065.85,2067.82,2069.73,2072.80,2073.80,2077.76,2079.71,2080.66,2080.69,2082.71,2082.71,2084.69,2084.71,2084.76,2086.65,2086.66,2088.67,2091.67,2093.64,2095.73,2097.64,2101.65,2101.84,2109.72,2111.77,2116.66,2118.73,2119.74,2123.69,2124.66,2126.70,2128.66,2136.67,2138.69,2148.66,2277.01,2283.00,2292.01,2293.92,2295.94,2296.01,2298.03,2298.96,2301.02,2303.01,2305.01,2309.00,2312.00,2314.99,2316.99,2319.00,2321.89,2323.00,2325.95,2328.86,2329.97,2331.89,2332.99,2333.89,2333.90,2336.97,2339.95,2342.89,2346.88,2351.86,2351.88,2353.91,2356.86,2356.97,2358.93,2361.86,2362.96,2364.81,2365.88,2366.96,2369.96,2372.79,2372.92,2377.89,2380.02,2380.81,2383.99,2385.79,2387.96,2390.76,2390.92,2394.81,2408.73,2411.86,2414.77,2414.82,2415.82,2416.90,2418.88,2423.61,2424.69,2427.70,2427.71,2427.78,2439.65,2440.83,2441.75,2442.69,2445.77,2446.76,2446.85,2448.79,2450.82,2454.78,2455.75,2466.74,2472.75,2474.76,2480.71,2483.71,2486.68,2491.72,2495.73,2502.65,2508.69,2511.69,2649.97,2650.95,2659.02,2661.02,2667.94,2667.99,2668.00,2673.99,2684.00,2687.95,2692.01,2696.97,2698.99,2703.96,2706.05,2706.97,2708.94,2712.91,2714.00,2715.89,2718.90,2719.88,2720.85,2721.95,2726.09,2727.86,2727.96,2732.94,2734.92,2735.91,2739.81,2743.88,2745.84,2745.95,2747.94,2750.71,2752.76,2753.86,2754.90,2756.94,2759.78,2759.82,2760.85,2761.90,2762.88,2763.82,2765.91,2768.75,2773.82,2780.90,2784.84,2792.62,2793.83,2793.88,2797.63,2798.63,2799.67,2799.77,2800.79,2801.64,2802.69,2805.81,2806.70,2807.85,2810.67,2817.83,2818.66,2819.75,2819.76,2823.62,2829.68,2830.60,2835.58,2837.71,2842.64,2844.65,2844.70,2850.70,2850.74,2854.73,2855.66,2858.73,2859.66,2864.71,2865.70,2869.70,2869.70,2873.68,2876.65,2896.66,3014.01,3018.02,3020.98,3025.02,3025.04,3029.03,3032.01,3034.96,3034.99,3039.03,3040.00,3042.04,3045.03,3049.00,3049.03,3051.97,3052.94,3056.93,3064.01,3064.93,3067.92,3071.93,3076.02,3078.94,3079.96,3080.90,3082.97,3087.91,3088.03,3088.91,3089.94,3095.02,3099.85,3101.91,3103.94,3105.96,3106.87,3107.95,3110.87,3114.88,3114.93,3115.83,3116.85,3116.92,3117.86,3118.86,3119.84,3121.85,3123.93,3125.89,3126.95,3129.89,3130.82,3133.86,3135.83,3137.76,3138.86,3143.85,3144.91,3145.81,3148.84,3151.77,3154.77,3156.76,3161.79,3161.86,3172.68,3176.84,3177.67,3183.71,3183.80,3189.69,3196.79,3201.78,3203.66,3205.77,3209.65,3223.65,3230.66,3232.67,3234.66,3243.65,3248.65,3388.04,3395.03,3402.02,3410.96,3412.00,3415.03,3419.00,3419.00,3422.94,3425.99,3426.94,3430.95,3433.97,3443.00,3445.99,3446.93,3450.88,3452.96,3453.95,3454.95,3455.94,3457.89,3458.92,3459.07,3459.86,3459.87,3460.94,3461.90,3462.88,3463.96,3464.94,3465.83,3465.88,3466.87,3470.87,3472.92,3474.93,3478.79,3479.89,3482.89,3483.87,3484.93,3485.89,3486.83,3487.84,3489.85,3491.82,3491.93,3492.78,3493.80,3501.85,3502.83,3506.82,3508.82,3510.81,3515.77,3515.78,3519.80,3524.76,3525.76,3528.80,3532.78,3535.73,3536.79,3546.75,3546.76,3548.75,3552.70,3552.73,3553.65,3558.68,3558.71,3559.73,3565.69,3575.70,3576.67,3584.65,3584.69,3589.70,3592.71,3596.67,3600.68,3601.67,3611.65,3738.03,3747.00,3756.02,3756.04,3763.01,3766.99,3772.03,3776.02,3778.02,3779.01,3781.03,3785.01,3787.94,3787.99,3793.06,3794.98,3801.02,3808.94,3811.97,3812.90,3818.86,3819.01,3825.95,3826.88,3830.94,3833.00,3837.88,3846.95,3849.77,3853.81,3853.92,3857.79,3860.87,3861.82,3867.80,3868.84,3873.84,3882.83,3884.78,3885.88,3889.77,3893.84,3897.85,3900.83,3904.76,3904.84,3907.78,3912.75,3919.71,3919.71,3925.75,3927.71,3935.68,3936.69,3947.67,3951.69,3953.70,3963.66,3967.66,3971.66,3976.66,4128.04,4134.01,4137.02,4139.96,4142.98,4149.01,4154.98,4159.02,4162.00,4167.01,4170.95,4173.94,4184.89,4188.95,4192.84,4196.82,4202.93,4209.94,4214.93,4218.89,4222.86,4226.87,4229.85,4230.86,4239.77,4240.87,4244.81,4249.76,4259.66,4264.73,4269.81,4273.70,4277.65,4281.74,4286.68
]

sim.set_instrument(
    name="testCAM-i",  # ECAM-r, LSSTg-g, LSSTi-i, delta-x, oddCAM-i, testCAM-i
    ZP=22.19,  # photometric magnitude zero point
    xmin=-1.75,  # minimum horizontal field-of-view coordinate in arcsec
    xmax=1.75,  # maximum
    ymin=-1.75,  # minimum vertical field-of-view coordinate in arcsec
    ymax=1.75,  # maximum
    time=obs_time,  # Nondecreasing observing epochs in days. None omits time; required for variability.
)

# Set instrument noise
sim.set_noise(
    instrument='testCAM-i',
    model='PoissonNoise',  # 'PoissonNoise', 'UniformGaussian' or 'NoNoise'
    Msb=22.8,  # Sky surface brightness in mag/arcsec^2; used by PoissonNoise.
    texp=200,  # Exposure time in seconds; must be positive for PoissonNoise
    sn=0.0,  # Signal-to-noise setting for UniformGaussian
    sigma=0.0,  # Noise standard deviation in pixel flux units for UniformGaussian
)


# ----------------------------------------------------------------------------
# ---------------------------- STEP 3: Lens model ----------------------------
# ----------------------------------------------------------------------------

# Lens redshift
sim.set_lens_redshift(redshift=0.77)

# Set the components of the mass model through set_lens_mass_model().
# Below, we call set_lens_mass_model multiple times, in order to set multiple components.
# The same component type, e.g. 'sie', can be set multiple times.
# Every physical parameter must be supplied explicitly on each mass-model call.
# Cases:
# - sie: Singular Isothermal Ellipsoid.
# - spemd: Softened Power-law Elliptical Mass Distribution.
# - external_shear: this has zero convergence, not a galaxy mass profile. A shear-only list is allowed but is not a usual strong lens.
# - pert: pixelated lensing potential read from a FITS grid.
# Note that each call to set_lens_mass_model sets a component. Beware of this when you run these scripts in a notebook.
# To reset the mass model components, you have to re-define the MoletInterface object.
sim.set_lens_mass_model(
    model='sie',
    theta_E=1.1,
    q=0.8,
    pa=35.0,
    x0=0.0,
    y0=0.0,
)

sim.set_lens_mass_model(
    model='external_shear',
    g=0.032,
    phi=-40.0,
    x0=0.0,
    y0=0.0,
)

# Set the components of the lens light profile.
# Light-profile type: 'gauss', 'sersic', or 'custom'.
# Every call appends one component to lens.light_profile[instrument],
# where instrument is an input to set_lens_light_profile (see the code below)
# and is set through set_instrument (see the code above).
sim.set_lens_light_profile(
    instrument='testCAM-i',
    profile='sersic',  # 'gauss', 'sersic', or 'custom'
	x0=0.0,  # Source-plane angular center in arcsec
	y0=0.0,
	pa=35.0,  # Position angle in degrees. VKL adds 90 degrees before rotating; compare conventions explicitly when importing another code's model.
	q=0.8,  # Dimensionless minor-to-major axis ratio; physical ellipses use 0<q<=1
	M_tot=21,  # Total component magnitude in this band. F=10**(-0.4*(M_tot-ZP)) is the integrated flux in electrons/s. The parser supplies ZP from the matching instrument entry. Fluxes, not magnitudes, add across components.
	r_eff=1.7,  # Positive angular scale in arcsec; see the model-specific definitions
	n=4,  # Dimensionless Sersic index
    mass_to_light={"upsilon": 2.0, "upsilon_exp": 0.2},
)

# For microlensing, use set_compact_mass_model():
"""
system.set_compact_mass_model(
    models=[
        {
            "type": "sersic",
            "pars": {
                "x0": 0.0,
                "y0": 0.0,
                "q": 0.8,
                "pa": 35.0,
                "r_eff": 1.0,
                "n": 4.0,
                "i_eff": 0.1,  # Compact convergence normalization at the effective radius, not surface brightness
            },
        }
    ],
)
"""


# ----------------------------------------------------------------------------
# --------------------------- STEP 4: Source model ---------------------------
# ----------------------------------------------------------------------------

# Set the source redshift
sim.set_source_redshift(2.03)

# Set the components of the source light profile.
# See the comments related to set_lens_light_profile above.
sim.set_source_light_profile(  # first component
    instrument="testCAM-i",
    profile="gauss",
    x0=-0.05,
	y0=0.05,
	pa=23.0,
	q=0.64,
	M_tot=25.0,
	r_eff=0.06,
)

sim.set_source_light_profile(  # second component
    instrument="testCAM-i",
    profile="gauss",
    x0=-0.2,
    y0=0.125,
    pa=0.0,
    q=1.0,
    M_tot=25.0,
    r_eff=0.08,
)

# If there's an AGN in the lensed host galaxy, use:
sim.set_point_source(
    x0=-0.05,
	y0=0.05,
    M_tot_unlensed=25.0,  # Total APPARENT point-source magnitude the unresolved source would have without gravitational magnification, used for the static macro-only image
    triangle_size=0.3,  # Largest initial triangle side in arcsec for point-source image finding
)


# ----------------------------------------------------------------------------
# ------------------------ STEP 5: Source variability ------------------------
# ----------------------------------------------------------------------------

# NOTE: Variability is allowed only for point-like sources

# WARNING regarding set_intrinsic_variability:
# the local DRW implementation currently uses only the shortest-wavelength
# band's mean and copies its curves into other bands.

# NOTE on set_extrinsic_variability:
# it affects the microlensing modulation of the point-like source

# NOTE on set_unmicrolensed_variability:
# Use this optional component for reverberating emission from a region assumed
# large enough that stellar microlensing is negligible, e.g. an idealized extended
# reprocessing region around an AGN. It remains subject to macro magnification and
# macro-image time delays. It is neither a static host profile nor a switch that
# disables microlensing for the entire source.

sim.set_intrinsic_variability(
    model="drw",
    N_in=1,
    absolute_i_mag=-23.0,
    mean_mag={"testCAM-i": 25.0},
    scale_factor=1.0,
)
```

```python
from pathlib import Path
import json

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize, TwoSlopeNorm
from astropy.io import fits

out = result.output_dir
run = out.parent
band = "testCAM-i"
mock = run / "mock_0000_0000"

images = json.loads((out / "multiple_images.json").read_text())
criticals = json.loads((out / "criticals.json").read_text())
caustics = json.loads((out / "caustics.json").read_text())
intrinsic = json.loads((out / f"{band}_LC_intrinsic.json").read_text())
continuous = json.loads((mock / f"{band}_LC_continuous.json").read_text())
sampled = json.loads((mock / f"{band}_LC_sampled.json").read_text())

image_ids = np.arange(len(images))
colors = plt.cm.tab10(image_ids % 10)

plt.rcParams.update({"figure.dpi": 110, "font.size": 11})
```

All FITS files

```python
for path in sorted(out.glob("*.fits")):
    data, header = fits.getdata(path, header=True)
    data = np.asarray(data, dtype=float)

    if data.ndim != 2 or not np.isfinite(data).any():
        continue

    has_extent = all(k in header for k in ("XMIN", "XMAX", "YMIN", "YMAX"))
    extent = (
        [float(header[k]) for k in ("XMIN", "XMAX", "YMIN", "YMAX")]
        if has_extent else None
    )

    finite = data[np.isfinite(data)]
    signed = path.name == "detA.fits" or "noise_realization" in path.name

    if signed:
        limit = max(float(np.percentile(np.abs(finite), 99.5)), 1e-12)
        norm = TwoSlopeNorm(vmin=-limit, vcenter=0, vmax=limit)
        cmap = "RdBu_r"
    else:
        low, high = np.percentile(finite, [0.5, 99.5])
        if high <= low:
            high = low + max(abs(low) * 0.01, 1e-12)
        norm = Normalize(vmin=low, vmax=high)
        cmap = "magma"

    fig, ax = plt.subplots(figsize=(6, 5), constrained_layout=True)
    im = ax.imshow(
        np.ma.masked_invalid(data),
        origin="lower", extent=extent,
        norm=norm, cmap=cmap, interpolation="nearest",
    )

    unit = "arcsec" if has_extent else "pixel"
    ax.set(
        title=path.stem,
        xlabel=f"x [{unit}]",
        ylabel=f"y [{unit}]",
    )

    label = header.get("BUNIT", "Native FITS values")
    if path.name == "detA.fits":
        label = "Jacobian determinant (dimensionless)"
    elif path.name == "lens_kappa_star_super.fits":
        label = r"$\kappa_\star$ (dimensionless)"

    fig.colorbar(im, ax=ax, label=label, extend="both")
    plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)

for ax, suffix, title in zip(
    axes,
    ["", "_ps_macro"],
    ["Extended components", "Including the macro point source"],
):
    noise = fits.getdata(out / f"{band}{suffix}_noise_realization.fits")
    sigma = fits.getdata(out / f"{band}{suffix}_sigma_map.fits")

    valid = np.isfinite(noise) & np.isfinite(sigma) & (sigma > 0)

    if valid.any():
        residual = noise[valid] / sigma[valid]
        ax.hist(residual, bins=60, density=True, alpha=0.7, label="Pixels")
        x = np.linspace(-5, 5, 400)
        ax.plot(
            x, np.exp(-x**2 / 2) / np.sqrt(2 * np.pi),
            color="black", label="Standard normal reference",
        )
        ax.legend()

    ax.set(xlabel="Noise / local sigma", ylabel="Density", title=title)

plt.show()
```
